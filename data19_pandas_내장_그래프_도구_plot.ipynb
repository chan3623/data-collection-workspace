# Pandas 

판다스(Pandas) 라이브러리는 자체적인 데이터프레임(df) 및 시리즈(s) 객체에 Matplotlib을 내장(df.plot())하고 있어, 데이터 전처리 후 즉시 빠르게 시각화할 수 있는 강력한 편의 기능을 제공한다.   
판다스 그래프는 plot() 메서드를 사용해 데이터프레임과 시리즈를 빠르게 시각화하는 기능이다.  
내부적으로 맷플롯립(Matplotlib)을 활용하므로 코드가 간결하다.

### 1. Pandas 시각화란? (정의)

* **Pandas 시각화**는 판다스의 데이터 구조인 **데이터프레임(DataFrame)** 및 **시리즈(Series)** 객체에 Matplotlib 기능을 내장(`df.plot()`)하여 **데이터 전처리 과정에서 즉시 그래프를 그릴 수 있도록 지원하는 고수준 편의 기능**입니다.
* 데이터 변환 후 별도의 복잡한 시각화 코드 작성 없이 단 한 줄의 코드로 탐색적 데이터 분석(EDA)을 진행할 수 있게 도와줍니다.



### 2. 주요 특징 (Features)

* **극강의 편의성 (한 줄 출력):** 데이터프레임 인덱스(Index)와 컬럼(Column)을 **X축, Y축, 범례(Legend)로 자동 매핑**해주므로 별도의 데이터 추출 작업이 필요 없습니다.
* **Matplotlib 호환성:** 내부적으로 Matplotlib을 기반으로 구동되기 때문에, Matplotlib의 축 객체(`ax`)나 세부 설정 명령어(`plt.title`, `plt.xlabel` 등)와 완벽하게 연동됩니다.
* **직관적인 `kind` 파라미터:** `kind='line'`, `'bar'`, `'scatter'` 등 인자 변경만으로 그래프 형태를 손쉽게 전환할 수 있습니다.
* **시계열 데이터 최적화:** 날짜/시간(Datetime) 인덱스를 자동으로 인식하여 X축 눈금을 보기 쉽게 최적화하여 표시해 줍니다.



### 3. 주요 `kind` 옵션 및 파라미터

```text
[Pandas .plot() 체계]
 ├── 기본 실행: df.plot(kind='...', x='col1', y='col2')
 └── 축약 실행: df.plot.line(), df.plot.bar(), df.plot.hist(), df.plot.scatter()

```

| `kind` 옵션 | 메서드 형태 | 주요 용도 및 특징 |
| --- | --- | --- |
| `'line'` | `df.plot.line()` | **기본값.** 시계열/연속형 데이터의 추세 파악 |
| `'bar'` / `'barh'` | `df.plot.bar()` | 범주형 데이터 비교 (수직/수평, `stacked=True`로 누적 막대 가능) |
| `'hist'` | `df.plot.hist()` | 연속형 변수의 구간별 도수분포 파악 (`bins` 옵션 활용) |
| `'box'` | `df.plot.box()` | 사분위수 기반 중앙값, 분포 범위 및 이상치(Outlier) 확인 |
| `'kde'` / `'density'` | `df.plot.kde()` | 데이터 분포의 부드러운 확률밀도 곡선 추정 |
| `'scatter'` | `df.plot.scatter()` | 두 연속형 변수 간의 상관관계 시각화 (`x`, `y` 지정 필수) |
| `'pie'` | `series.plot.pie()` | 단일 Series의 카테고리별 비중/비율(%) 확인 |
| `'hexbin'` | `df.plot.hexbin()` | 대용량 데이터의 2D 고밀도 집계 시각화 |


### 4. 핵심 기능 및 차트 종류

(1) Pandas 시각화 목적

```python

 ├── 1. 시간 흐름/추세 파악 ───► 선그래프 (plot.line), 면적 그래프 (plot.area)
 ├── 2. 항목 간 크기 비교 ────► 막대그래프 (plot.bar / plot.barh)
 ├── 3. 변수 간 상관관계 ────► 산점도 (plot.scatter), 고밀도 육각 산점도 (plot.hexbin)
 ├── 4. 분포 및 이상치 확인 ──► 히스토그램 (plot.hist), 박스플롯 (plot.box), 커널 밀도 (plot.kde)
 ├── 5. 전체 대비 비율 확인 ──► 파이 차트 (plot.pie)
 └── 6. 다변량/특수 시각화 ───► 산점도 행렬 (scatter_matrix), 평행 좌표계 (parallel_coordinates), 앤드류스 곡선 (andrews_curves)
```

**(2) Pandas 주요 메서드**

| **목적** | **주요 메서드 / 그래프 종류** | **설명** |
| --- | --- | --- |
| **추세 파악** | `df.plot.line()`, `df.plot.area()` | 시계열 데이터 추세 및 각 변수의 누적 면적 변화 시각화 |
| **크기 비교** | `df.plot.bar()`, `df.plot.barh()` | 범주형 데이터 비교 (수직/수평, `stacked=True`로 누적 막대 가능) |
| **상관 관계** | `df.plot.scatter()`, `df.plot.hexbin()` | 두 연속형 변수 간 산점도 및 많은 양의 데이터 밀도를 보여주는 육각형 Binning 차트 |
| **분포/이상치** | `df.plot.hist()`, `df.plot.box()`, `df.plot.kde()` | 도수분포, 사분위수 기반 박스플롯, 확률밀도 곡선 파악 |
| **비율 확인** | `series.plot.pie()` | 단일 Series 데이터의 카테고리별 비율(%) 시각화 |
| **다변량 분석** | `pd.plotting.scatter_matrix()`, `pd.plotting.parallel_coordinates()` | 모든 변수 간의 산점도 행렬 일괄 출력 및 다차원 데이터의 집단별 패턴/특성 비교 |

### 5. 장점 및 단점

#### ⭕ 장점

1. **빠른 EDA 속도:** 데이터 가공 후 곧바로 `.plot()`을 붙여 데이터의 구조, 분포, 이상치를 즉시 파악할 수 있습니다.
2. **자동 인덱스 처리:** 인덱스(Index)를 X축으로 자동 지정해 주어 시계열 데이터 분석 시 가독성이 뛰어납니다.
3. **가볍고 간단함:** 추가 라이브러리를 임포트(import)하지 않고 Pandas만으로 시각화까지 한 번에 끝낼 수 있습니다.

#### ❌ 단점

1. **커스텀 제한:** 섬세한 디자인이나 복잡한 레이아웃을 다루기에는 한계가 있어 결국 Matplotlib 코드를 섞어 써야 합니다.
2. **통계 시각화 부재:** Seaborn처럼 회귀선 자동 생성이나 카테고리별 복잡한 분포 비교 기능은 제공하지 않습니다.



> **"데이터 가공과 시각화를 하나로 이어주는 탐색적 데이터 분석(EDA)의 패스파인더"**
> 데이터 전처리 중간 단계에서 **"데이터가 어떻게 생겼지?" 하고 빠르게 눈으로 확인하고 싶을 때** 가장 효과적인 도구입니다.

## 1.2 수직 막대그래프(Bar Plot)
- 범주형/그룹별 데이터의 값을 비교한다.

````python
# 평균을 보기 위해 bmi를 분위수별로 나눠서 수직막대그래프로 시각화
df['bmi_bin'] = pd.qcut(df['bmi'], 5)
print(df.head(1))

bmi_group = df.groupby('bmi_bin')['target'].mean()   # select mean(target) from diabetes group by df
bmi_group.plot(kind='bar', title='막대그래프: BMI 구간별 평균 당뇨 진행도', figsize=(8, 4))

plt.ylabel("평균 진행도")
plt.xlabel("BMI 분위 구간")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
````

## 1.3 수평 막대그래프(Barh Plot)
- 범주형/그룹별 데이터의 값을 비교한다.

````python
# 평균을 보기 위해 bmi를 분위수별로 나눠서 수평막대그래프로 시각화
df['bmi_bin'] = pd.qcut(df['bmi'], 5)
#print(df.head(1))

bmi_group = df.groupby('bmi_bin')['target'].mean()   # select mean(target) from diabetes group by df
bmi_group.plot(kind='barh', title='막대그래프: BMI 구간별 평균 당뇨 진행도', figsize=(8, 4))

plt.ylabel("평균 진행도")
plt.xlabel("BMI 분위 구간")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()

# target 값이 어느 구간에 집중되어 있는지 보여줍니다. 주로 70~150 사이에 데이터가 많이 분포하며 오른쪽으로 긴 꼬리를 갖는 형태를 띱다.
````

## 1.4 히스토그램(Histogram)
- 연속형 변수의 구간별 빈도(분포)를 확인한다.

````python
# target 변수 분포 확인
df['target'].plot(kind='hist', bins=20, color='skyblue', edgecolor='black',
                  title='히스토그램: 당뇨 진행도 분포', figsize=(8, 4))
plt.xlabel("당뇨 진행도")
plt.grid(True)
plt.show()

# .plot(kind='hist') 로 연속형 변수 분포 확인
# target값이 특정 범위에 집중되어 있음
````

## 1.5 박스플롯(Box)
- 데이터의 중앙값, 사분위수(IQR), 이상치(Outlier)를 한눈에 파악한다.

````python
# 여러 특성의 분포를 한 번에 박스플롯으로 보기
df[['bmi', 'bp', 's1', 's2', 's3', 's4', 's5']].plot(kind='box', figsize=(10, 6),
                                                              title='박스플롯: 주요 변수들의 분포')
plt.grid(True)
plt.show()

# 여러 열에 대한 분포, 중앙값, 이상치를 한 번에 비교
# .plot(kind='box')는 수치형 데이터 요약에 유용함.
# 가운데 50%는 중앙값을 의미한다.
# 박스 크기가 크면 데이터가 넓게 퍼져 있음을 의미하고, 작으면 집중되여 있음을 의미한다.
# 중앙값(50%) 선 위치로 각 변수의 중앙값 차이를 알 수 있다.
# 이상치가 많으면 해당 변수에 특이한 값들이 존재한다는 뜻이다.
````

## 1.6  커널 (밀도 그래프 'kde' / 'density')
 - 히스토그램을 부드러운 곡선 형태로 정규화하여 확률 밀도를 시각화 한다.

````python
df["bmi"].plot(
    kind="kde", figsize=(8, 4), title="KDE: BMI 변수 밀도 그래프")

plt.grid(True)
plt.show()

# 데이터 분포의 대략적인 형태와 정규성(종 모양 여부)을 곡선으로 파악할 수 있다.
````

## 1.7  면적 그래프(area)
- 선그래프 아래쪽 면적을 색으로 채워 양적 변화를 강조 한다.

````python
df[["s1", "s2"]].head(20).plot(
    kind="area",
    stacked=False,
    alpha=0.5,
    figsize=(8, 4),
    title="면적 그래프: s1, s2 변화",
)

plt.grid(True)
plt.show()

# 선 아래 영역의 크기를 통해 두 변수의 누적 정도나 변화량을 쉽게 비교할 수 있다.

````

## 1.8 파이 그래프(pie)
- 범주 비율을 원형으로 나타낸다. (Series 전용).

````python
# target 값을 3개 구간(저, 중, 고)으로 범주화 후 비율 시각화
target_group = pd.qcut(df["target"], q=3, labels=["하위", "중위", "상위"]).value_counts()


target_group.plot(kind="pie", autopct="%.1f%%", figsize=(6, 6), title="파이 그래프: 당뇨병 진행도 구간 비율")


plt.ylabel("")  # y축 라벨 제거
plt.show()

# 데이터가 지정한 각 범주에 몇 퍼센트(%)씩 분배되어 있는지 비율을 명확하게 보여준다.
````

## 1.9  산점도(Scatter Plot)
- 두 연속형 변수 간의 관계(상관관계)를 점으로 표현한다 (x, y 지정 필수).

````python
# 산점도는 kind='scatter'로 x, y 축을 명시해야 함
df.plot(kind='scatter', x='bmi', y='target',
        title='산점도: BMI vs 당뇨 진행도', figsize=(8, 4), color='purple')
plt.grid(True)
plt.show()

# BMI가 높을 수록 target도 증가하는 경향을 시각화
# .plot(kind='scatter')는 변수 간 관계 확인에 적합

# BMI가 높을수록 target(당뇨 진행도)도 증가하는 우상향 경향을 보입니다. 두 변수 사이에 양의 상관관계가 있음을 알 수 있다.
````

## 1.10 고밀도 산점도 그래프(hexbin)
- 데이터가 많아 점이 겹칠 때, 육각형 셀(2D Binning)의 색상 진하기로 밀도를 표현한다.

````python
df.plot(
    kind="hexbin",
    x="bmi",
    y="target",
    gridsize=15,
    cmap="Blues",
    figsize=(8, 4),
    title="고밀도 산점도: BMI vs 당뇨 진행도",
)


plt.show()

# 색이 짙은 육각형 영역에 데이터(환자) 샘플이 가장 많이 밀집되어 있음을 보여준다.
````
