```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.DataFrame(columns=['x', 'y'])
```

```python
df.loc[0] = [3, 1]
df.loc[1] = [4, 1]
df.loc[2] = [3, 2]
df.loc[3] = [4, 2]
df.loc[4] = [10, 5]
df.loc[5] = [10, 6]
df.loc[6] = [11, 5]
df.loc[7] = [11, 6]
df.loc[8] = [15, 1]
df.loc[9] = [15, 2]
df.loc[10] = [16, 1]
df.loc[11] = [16, 2]
```

```python
df.head(20)
```

```python
sns.lmplot(
    x='x',
    y='y',
    data=df,
    fit_reg=False,
    scatter_kws={"s": 200}
)

plt.title("kmean plot")

plt.xlabel('x')
plt.ylabel('y')
```

```python
data_points = df.values
```

```python
Kmeans = KMeans(n_clusters=3).fit(data_points)
```

```python
Kmeans.labels_
```

```python
Kmeans.cluster_centers_
```

```python
df['cluster_id'] = Kmeans.labels_
```

```python
df.head(12)
```

```python
sns.lmplot(x='x', y='y', data=df, fit_reg=False, scatter_kws={"s" : 150}, hue="cluster_id")

plt.title("after kmean clustering")
```
