```python
scores = np.array([
    [80, 90, 70],
    [60, 85, 75],
    [90, 95, 100],
    [70, 65, 80],
    [85, 75, 90]
])
```

```python
# 학생별 총점 구하기


print(scores.sum(axis=1))
```

```python
# 학생별 평균 구하기

print(scores.mean(axis=1))
```

```python
# 과목별 평균 구하기

print(scores.mean(axis=0))
```

```python
# 평균이 80점 이상인 학생만 출력

print(scores[scores.mean(axis=1) >= 80])
```

```python
sales = np.array([
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [20, 30, 40, 50]
])
```

```python
# 2번째 매장의 데이터 출력

print(sales[1])
```

```python
# 3주차 판매량만 출력

print(sales[:,2])
```

```python
# 전체 판매량 중 40 이상인 값만 출력

print(sales[sales >= 40])
```

```python
# 평균보다 높은 판매량 찾기

print(sales[sales > np.mean(sales)])
```

```python
stock = np.array([
    [
        [10,20,30],
        [15,25,35]
    ],
    [
        [20,30,40],
        [25,35,45]
    ]
])
```

```python
# 첫 번째 상품 전체 출력

print(stock[0,0])
```

```python
# 모든 상품의 2번째 월 재고 출력

print(stock[:,:,1])
```

```python
# 전체 재고 합계
print(np.sum(stock))
```

```python
# 상품별 총 재고

# print(np.sum(stock[0]), np.sum(stock[1]))
print(stock.sum(axis=0))
```
